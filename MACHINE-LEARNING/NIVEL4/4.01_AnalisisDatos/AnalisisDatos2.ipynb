# Estadística Descriptiva

## Medidas de Tendencia Central

* Principalmente son los siguiente:
  - ***Promedio o Media***: sumar todo y dividir entre el total de elementos.
  - ***Mediana***: tenemos que ordenar y seleccionar el de la posición central, si hay 2 empatados, se saca el promedio.
  - ***Moda***: valor que se repite más en nuestra muestra.
* Estas medidas intentan responder: *“¿Cuál es el valor representativo o central de los datos?”*.
### ***Media (promedio)***
* Se calcula sumando todos los valores y dividiendo entre el número de observaciones.
$$\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i$$
* La media intenta: *“balancear”* todos los datos.
* Es MUY sensible a valores extremos.
### ***Mediana***
* La mediana es: el valor que queda exactamente en el centro cuando ordenas los datos.
* ***NO se afecta mucho por outliers, es robusta***.
* Mucho más representativa que la media.
### ***Moda***
* La moda es: el valor que más se repite.
* Un dataset puede ser:
  - `unimodal` → una moda.
  - `bimodal` → dos modas.
  - `multimodal`.
  - `no modal`, si no hay valores que se repiten.
* * ***Distribución Simétrica***
  - mediana=media aproximadamente.
* ***Distribución Sesgada Positivamente (cola hacia la derecha)***
  - media > mediana.
* ***Distribución Sesgada negativamente (cola hacia la izquierda.)***
  - mediana > media.

## Medidas de Dispersión

* ***¿Qué tan separados están?”***
* Primera medida: `Rango`.
* La podemos calcular como $\text{Rango}= x_{\text{max}} - x_{\text{min}}$.
* Si tenemos estos datos `5, 7, 8, 9, 100`, su rango sería `100-5 = 95`.
* El problema del rango, es que ignora todos los datos del centro.
* Entonces aparece: `Varianza` $\sigma^2$ y `desviación estándar` $\sigma$ que ***estas sí consideran TODOS los datos***.
* La `desviación estándar` mide: ***qué tan lejos están los datos respecto a la media***.
* Mayor desviación estándar significa: los datos están más dispersos respecto al centro.
* La varianza mide el promedio de las distancias cuadradas respecto a la media.
$$\sigma^2 = \frac{1}{n}\sum_{i=1}^{n} (x_i - \mu)^2$$
* Estas son al cuadrado porque si pudieran salir valores negativos de las restas, y algunas podrían cancelarse entre si.
* Por otro lado la `desviación estándar` es simplemente su raíz cuadrada $\sigma = \sqrt{\sigma^2}$.
* Se usa más este porque vuelve a las unidades originales.
* La `desviación estándar` es algo así como: *“la distancia típica de los datos respecto a la media”*
* En un dataset donde la $\bar{x}=100$ pero sus desviaciones sean $\sigma_A=2$ y $\sigma_B=2$, podríamos entender:
  - El que tiene el dataset más estable o concentrado sería el primero debido a su pequeña desviación.
  - Mientra el otro tendría una variabilidad más grande y por lo tanto mayor probabilidad de encntrar `outliers`.

```python
from sklearn.datasets import load_iris
import pandas as pd

### Cargar dataset
iris = load_iris()

### Convertir a DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

### Agregar la clase
df["species"] = iris.target

print(df.head())
```

```python
### Media
print("Media:\n", df.mean())
### Media
print("\nMediana:\n", df.median())
### Moda
print("\nModa:\n", df.median())
### Varianza
print("\nVarianza:\n", df.var())
### Desviación
print("\nStd:\n", df.std())

### Resumen
print("\nResumen:\n", df.describe())
```
